# Reproduce topology-PINO paper figures

This notebook reads evaluator outputs only; it does not hide data generation or metric calculations.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

plt.style.use("seaborn-v0_8-whitegrid")
RESULTS = Path("../results")
```

```python
tables = []
for case in ("case30", "case57", "case118", "case300"):
    path = RESULTS / case / "pino" / "summary_metrics.csv"
    if path.exists():
        frame = pd.read_csv(path)
        row = frame.loc[frame.scope.eq("all_unseen_topologies")].copy()
        row["case"] = case
        tables.append(row)
summary = pd.concat(tables, ignore_index=True) if tables else pd.DataFrame()
summary
```

```python
if not summary.empty:
    fig, ax = plt.subplots(figsize=(5.5, 3.4), constrained_layout=True)
    ax.plot(summary["n_bus_mean"], summary["inference_latency_ms_mean"], "o-", label="GFNO-PINO")
    ax.plot(summary["n_bus_mean"], summary["classical_ac_latency_ms_mean"], "s--", label="AC-OPF")
    ax.set(xlabel="Bus count", ylabel="Latency (ms)", yscale="log")
    ax.legend()
    fig.savefig(RESULTS / "scalability_latency.png", dpi=300)
```
